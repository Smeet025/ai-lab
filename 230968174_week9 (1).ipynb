```python
# Representing facts (observed events)
facts = [
    "MultipleFailedLogins",
    "LoginFromUnknownIP",
    "AccessToSensitiveFiles",
    "AdminPrivileges"
]
```

```python
# Representing rules (knowledge base)
rules = [
    ("MultipleFailedLogins ∧ LoginFromUnknownIP → SuspiciousLogin"),
    ("SuspiciousLogin ∧ AdminPrivileges → HighRiskIntrusion"),
    ("HighRiskIntrusion → IntrusionDetected"),
    ("AccessToSensitiveFiles ∧ ¬AdminPrivileges → IntrusionDetected")
]
```

## Part 1: Cybersecurity Threat Detection

```python
def eliminate_implications(rule):
    # Eliminate implications: A → B becomes ¬A ∨ B
    lhs, rhs = rule.split("→")
    lhs = lhs.strip()
    rhs = rhs.strip()
    return f"¬({lhs}) ∨ {rhs}"


cnf_rules = [eliminate_implications(rule) for rule in rules]
print("CNF Rules:")
for rule in cnf_rules:
    print(rule)
```

```python
# To apply resolution, we'll negate the query (¬IntrusionDetected) and add it to the set of clauses.

query = "IntrusionDetected"
negated_query = f"¬{query}"

print(f"Negated query: {negated_query}")
```

```python
# Resolution rule implementation
def resolution(clauses):
    
    new_clauses = set(clauses)
    resolved = False

    # Apply resolution between all pairs of clauses
    for clause1 in clauses:
        for clause2 in clauses:
            if clause1 != clause2:
                # Check if there's a complementary literal between clause1 and clause2
                if "¬" + clause1.split(" ∨ ")[0] in clause2 or clause1.split(" ∨ ")[0] in clause2:
                    new_clause = f"{clause1.split(' ∨ ')[1]} ∨ {clause2.split(' ∨ ')[1]}"
                    new_clauses.add(new_clause)
                    resolved = True

    # If we derived an empty clause, the resolution is successful
    if "" in new_clauses:
        print("Resolution successful: IntrusionDetected = TRUE")
    else:
        print("Resolution unsuccessful: No intrusion detected.")
    return new_clauses


cnf_clauses = [
    "¬MultipleFailedLogins ∨ ¬LoginFromUnknownIP ∨ SuspiciousLogin",
    "¬SuspiciousLogin ∨ ¬AdminPrivileges ∨ HighRiskIntrusion",
    "¬HighRiskIntrusion ∨ IntrusionDetected",
    "¬AccessToSensitiveFiles ∨ AdminPrivileges ∨ IntrusionDetected"
]

cnf_clauses.append(negated_query)

resolution(cnf_clauses)
```

## Part 2: Wastewater Treatment Decision

```python
# Representing facts (sensor readings)
facts_wastewater = [
    "HighBOD",
    "LowDO",
    "HighTurbidity",
    "ToxicChemicals",
    "pHImbalance"
]
```

```python
# Representing rules (domain knowledge)
rules_wastewater = [
    ("HighBOD ∧ LowDO → OrganicPollution"),
    ("HighTurbidity ∨ ToxicChemicals → Contamination"),
    ("OrganicPollution ∧ Contamination → SeverePollution"),
    ("SeverePollution → TreatmentRequired"),
    ("pHImbalance → TreatmentRequired")
]
```

```python
# Convert all wastewater rules into CNF format by eliminating implications
cnf_wastewater_rules = [eliminate_implications(rule) for rule in rules_wastewater]
print("CNF Wastewater Rules:")
for rule in cnf_wastewater_rules:
    print(rule)
    
```

```python
# Negate the query (TreatmentRequired) for resolution
query_wastewater = "TreatmentRequired"
negated_query_wastewater = f"¬{query_wastewater}"

print(f"Negated query for wastewater: {negated_query_wastewater}")

# Sample CNF clauses for wastewater
cnf_wastewater_clauses = [
    "¬HighBOD ∨ ¬LowDO ∨ OrganicPollution",
    "¬HighTurbidity ∨ ¬ToxicChemicals ∨ Contamination",
    "¬OrganicPollution ∨ ¬Contamination ∨ SeverePollution",
    "¬SeverePollution ∨ TreatmentRequired",
    "¬pHImbalance ∨ TreatmentRequired"
]


cnf_wastewater_clauses.append(negated_query_wastewater)


resolution(cnf_wastewater_clauses)
```
