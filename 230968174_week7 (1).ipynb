# Tic-Tac-Toe (Minimax + Alpha–Beta)

```python
# Initialize the board and define basic helper functions for the game.
nodes_minimax = 0
nodes_ab = 0

def initial_state():
    return [' ' for _ in range(9)]

def print_board(state):
    print("\n")
    for i in range(3):
        print(state[3*i], "|", state[3*i+1], "|", state[3*i+2])
        if i < 2:
            print("--+---+--")
    print("\n")
```

```python
# Define possible actions, apply moves, and determine which player's turn it is.
def actions(state):
    return [i for i in range(9) if state[i] == ' ']

def result(state, action, player):
    new_state = state.copy()
    new_state[action] = player
    return new_state

def player(state):
    if state.count('X') <= state.count('O'):
        return 'X'
    else:
        return 'O'
```

```python
# Implement winning conditions, terminal test, and utility evaluation.
def winner(state):
    win_positions = [
        [0,1,2],[3,4,5],[6,7,8],
        [0,3,6],[1,4,7],[2,5,8],
        [0,4,8],[2,4,6]
    ]
    for pos in win_positions:
        if state[pos[0]] == state[pos[1]] == state[pos[2]] != ' ':
            return state[pos[0]]
    return None

def terminal(state):
    return winner(state) is not None or ' ' not in state

def utility(state):
    win = winner(state)
    if win == 'X':
        return 1
    elif win == 'O':
        return -1
    else:
        return 0
```

```python
# Implement the Minimax algorithm along with node counting.
def max_value(state):
    global nodes_minimax
    nodes_minimax += 1

    if terminal(state):
        return utility(state)

    v = float('-inf')
    for action in actions(state):
        v = max(v, min_value(result(state, action, 'X')))
    return v

def min_value(state):
    global nodes_minimax
    nodes_minimax += 1

    if terminal(state):
        return utility(state)

    v = float('inf')
    for action in actions(state):
        v = min(v, max_value(result(state, action, 'O')))
    return v

def minimax_decision(state):
    best_score = float('-inf')
    best_move = None
    for action in actions(state):
        value = min_value(result(state, action, 'X'))
        if value > best_score:
            best_score = value
            best_move = action
    return best_move
```

```python
# Improve the Minimax algorithm using Alpha-Beta pruning to reduce search space.
def max_value_ab(state, alpha, beta):
    global nodes_ab
    nodes_ab += 1

    if terminal(state):
        return utility(state)

    v = float('-inf')
    for action in actions(state):
        v = max(v, min_value_ab(result(state, action, 'X'), alpha, beta))
        if v >= beta:
            return v
        alpha = max(alpha, v)
    return v

def min_value_ab(state, alpha, beta):
    global nodes_ab
    nodes_ab += 1

    if terminal(state):
        return utility(state)

    v = float('inf')
    for action in actions(state):
        v = min(v, max_value_ab(result(state, action, 'O'), alpha, beta))
        if v <= alpha:
            return v
        beta = min(beta, v)
    return v

def alphabeta_decision(state):
    best_score = float('-inf')
    best_move = None
    alpha = float('-inf')
    beta = float('inf')

    for action in actions(state):
        value = min_value_ab(result(state, action, 'X'), alpha, beta)
        if value > best_score:
            best_score = value
            best_move = action
        alpha = max(alpha, best_score)
    return best_move
```

```python
###### Create the main game loop for Human vs AI gameplay.
def play_game():
    global nodes_minimax, nodes_ab
    state = initial_state()

    while not terminal(state):
        print_board(state)

        if player(state) == 'X':
            print("AI is thinking...")
            move = alphabeta_decision(state)
            state = result(state, move, 'X')
            print("AI chose position:", move)
        else:
            move = int(input("Enter position (0-8): "))
            if move in actions(state):
                state = result(state, move, 'O')
            else:
                print("Invalid move!")
                continue

    print_board(state)

    if winner(state):
        print(winner(state), "wins!")
    else:
        print("Draw!")

    print("Nodes evaluated using Minimax:", nodes_minimax)
    print("Nodes evaluated using Alpha-Beta:", nodes_ab)

play_game()
```

# Car Overtaking Game

```python
# Initialize road size, counters, and starting positions.
nodes_minimax_car = 0
nodes_ab_car = 0

ROAD_END = 6

def initial_state_car():
    return (0, 1)
```

```python
# Define valid moves and update positions after each move.
def actions_car(state, current_player):
    x, o = state
    moves = []
    
    pos = x if current_player == 'X' else o
    
    for step in [1, 2]:
        new_pos = pos + step
        if new_pos <= ROAD_END:
            if current_player == 'X' and new_pos != o:
                moves.append(step)
            if current_player == 'O' and new_pos != x:
                moves.append(step)
    return moves

def result_car(state, move, current_player):
    x, o = state
    if current_player == 'X':
        return (x + move, o)
    else:
        return (x, o + move)
```

```python
# Implement terminal condition and utility evaluation for the game.
def terminal_car(state):
    x, o = state
    return x == ROAD_END or o == ROAD_END

def utility_car(state):
    x, o = state
    if x == ROAD_END:
        return 1
    elif o == ROAD_END:
        return -1
    else:
        return 0
```

```python
# Apply the Minimax algorithm with node tracking for comparison.
def max_value_car(state):
    global nodes_minimax_car
    nodes_minimax_car += 1

    if terminal_car(state):
        return utility_car(state)

    v = float('-inf')
    for move in actions_car(state, 'X'):
        v = max(v, min_value_car(result_car(state, move, 'X')))
    return v

def min_value_car(state):
    global nodes_minimax_car
    nodes_minimax_car += 1

    if terminal_car(state):
        return utility_car(state)

    v = float('inf')
    for move in actions_car(state, 'O'):
        v = min(v, max_value_car(result_car(state, move, 'O')))
    return v

def minimax_decision_car(state):
    best_move = None
    best_value = float('-inf')
    for move in actions_car(state, 'X'):
        value = min_value_car(result_car(state, move, 'X'))
        if value > best_value:
            best_value = value
            best_move = move
    return best_move
```

```python
# Optimize search using Alpha-Beta pruning to reduce unnecessary exploration.
def max_value_ab_car(state, alpha, beta):
    global nodes_ab_car
    nodes_ab_car += 1

    if terminal_car(state):
        return utility_car(state)

    v = float('-inf')
    for move in actions_car(state, 'X'):
        v = max(v, min_value_ab_car(result_car(state, move, 'X'), alpha, beta))
        if v >= beta:
            return v
        alpha = max(alpha, v)
    return v

def min_value_ab_car(state, alpha, beta):
    global nodes_ab_car
    nodes_ab_car += 1

    if terminal_car(state):
        return utility_car(state)

    v = float('inf')
    for move in actions_car(state, 'O'):
        v = min(v, max_value_ab_car(result_car(state, move, 'O'), alpha, beta))
        if v <= alpha:
            return v
        beta = min(beta, v)
    return v

def alphabeta_decision_car(state):
    best_move = None
    best_value = float('-inf')
    alpha = float('-inf')
    beta = float('inf')

    for move in actions_car(state, 'X'):
        value = min_value_ab_car(result_car(state, move, 'X'), alpha, beta)
        if value > best_value:
            best_value = value
            best_move = move
        alpha = max(alpha, best_value)

    return best_move
```

```python
# Run the full simulation and display moves along with final results.
def play_car_game():
    state = initial_state_car()
    current = 'X'

    while not terminal_car(state):
        print("Current Positions:", state)

        if current == 'X':
            move = alphabeta_decision_car(state)
            print("X moves:", move)
            state = result_car(state, move, 'X')
            current = 'O'
        else:
            move = minimax_decision_car(state)
            print("O moves:", move)
            state = result_car(state, move, 'O')
            current = 'X'

    print("Final Positions:", state)

    if utility_car(state) == 1:
        print("X wins!")
    else:
        print("O wins!")

    print("Minimax nodes:", nodes_minimax_car)
    print("Alpha-Beta nodes:", nodes_ab_car)

play_car_game()
```
