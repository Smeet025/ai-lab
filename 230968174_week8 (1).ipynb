```python
from typing import Dict
```

```python
# This will take sensor inputs and decide whether to turn lights ON or OFF
def smart_lighting_agent(percepts: Dict[str, bool]) -> str:
    # Manual override has highest priority, ignore all other rules
    if percepts.get("ManualOverride", False):
        return "Manual Override Active: No change"
    # Turn light ON if it's dark and motion is detected
    if percepts.get("IsDark", False) and percepts.get("MotionDetected", False):
        return "Turn Light ON"
    # Turn light OFF if no motion is detected
    if not percepts.get("MotionDetected", False):
        return "Turn Light OFF"
    # Default action if no rule applies
    return "Keep Light as is"
```

```python
# Test the smart lighting agent with some example percepts
# Let's see what the agent does in different situations
test_percepts = [
    {"MotionDetected": True, "IsDark": True, "ManualOverride": False},
    {"MotionDetected": False, "IsDark": True, "ManualOverride": False},
    {"MotionDetected": True, "IsDark": False, "ManualOverride": False},
    {"MotionDetected": True, "IsDark": True, "ManualOverride": True},
]

for p in test_percepts:
    action = smart_lighting_agent(p)
    print(f"Percepts: {p} -> Action: {action}")
```

```python
# Define a forward-chaining medical diagnosis agent
# This will suggest possible diseases based on symptoms
def medical_diagnosis_agent(percepts: Dict[str, bool]) -> Dict[str, bool]:
    knowledge = {}
    # Fever AND Cough means Flu
    if percepts.get("Fever", False) and percepts.get("Cough", False):
        knowledge["Flu"] = True
    # Fever AND Cough AND ChestPain means Pneumonia
    if percepts.get("Fever", False) and percepts.get("Cough", False) and percepts.get("ChestPain", False):
        knowledge["Pneumonia"] = True
    # Pneumonia requires hospitalization
    if knowledge.get("Pneumonia", False):
        knowledge["HospitalizationRequired"] = True
    return knowledge
```

```python
# Test the medical diagnosis agent with example symptoms
# Let's see what diseases and actions are inferred
symptoms = [
    {"Fever": True, "Cough": True, "ChestPain": False},
    {"Fever": True, "Cough": True, "ChestPain": True},
    {"Fever": False, "Cough": True, "ChestPain": True},
]

for s in symptoms:
    diagnosis = medical_diagnosis_agent(s)
    print(f"Symptoms: {s} -> Diagnosis: {diagnosis}")
```

```python
# Define traffic violation agent
# This checks for speeding and signal violations with emergency vehicle exceptions
def traffic_violation_agent(percepts: Dict[str, bool]) -> Dict[str, bool]:
    violations = {"SpeedViolation": False, "SignalViolation": False}
    
    # If emergency vehicle, ignore all violations
    if percepts.get("EmergencyVehicle", False):
        return {"SpeedViolation": False, "SignalViolation": False, "Note": "Emergency Vehicle: No violations"}
    
    # Speed above limit triggers speed violation
    if percepts.get("SpeedAboveLimit", False):
        violations["SpeedViolation"] = True
    # Running red light triggers signal violation
    if percepts.get("SignalRed", False) and percepts.get("VehicleCrossedLine", False):
        violations["SignalViolation"] = True
    
    return violations
```

```python
# Test the traffic violation agent with different scenarios
# Check how the agent responds to normal vs emergency vehicles
traffic_events = [
    {"SpeedAboveLimit": True, "SignalRed": False, "VehicleCrossedLine": False, "EmergencyVehicle": False},
    {"SpeedAboveLimit": True, "SignalRed": True, "VehicleCrossedLine": True, "EmergencyVehicle": False},
    {"SpeedAboveLimit": True, "SignalRed": True, "VehicleCrossedLine": True, "EmergencyVehicle": True},
]

for e in traffic_events:
    result = traffic_violation_agent(e)
    print(f"Percepts: {e} -> Violations: {result}")
```

```python
# Define a cybersecurity logical agent
# This infers suspicious activity and triggers alerts
def cybersecurity_agent(percepts: Dict[str, bool]) -> str:
    suspicious = False
    high_risk = False
    
    # Multiple failed logins from unknown IPs are suspicious
    if percepts.get("MultipleFailedLogins", False) and percepts.get("LoginFromUnknownIP", False):
        suspicious = True
    
    # Suspicious login with admin privileges is high risk
    if suspicious and percepts.get("AdminPrivileges", False):
        high_risk = True
    
    # Decide action
    if high_risk:
        return "Trigger ALERT: High Risk Intrusion"
    if suspicious:
        return "Suspicious login detected"
    return "No issues detected"
```

```python
# Test cybersecurity agent
# Simulate different login scenarios to check agent response
login_events = [
    {"MultipleFailedLogins": True, "LoginFromUnknownIP": True, "AdminPrivileges": False},
    {"MultipleFailedLogins": True, "LoginFromUnknownIP": True, "AdminPrivileges": True},
    {"MultipleFailedLogins": False, "LoginFromUnknownIP": False, "AdminPrivileges": True},
]

for event in login_events:
    action = cybersecurity_agent(event)
    print(f"Percepts: {event} -> Action: {action}")
```

```python
# Define warehouse robot agent
# The robot decides movement based on obstacles, humans, and fragile items
def warehouse_robot_agent(percepts: Dict[str, bool]) -> str:
    # Stop if obstacle ahead
    if percepts.get("ObstacleAhead", False):
        if percepts.get("CarryingFragileItem", False):
            return "Slow Down due to fragile item and obstacle"
        return "Stop: Obstacle ahead"
    
    # Reduce speed if human nearby
    if percepts.get("HumanNearby", False):
        return "Reduce Speed: Human nearby"
    
    # Move forward if path is clear and no human nearby
    if percepts.get("PathClear", False) and not percepts.get("HumanNearby", False):
        return "Move Forward"
    
    # Default action
    if percepts.get("DestinationReached", False):
        return "Destination reached: Stop"
    
    return "Wait: Conditions unclear"
```

```python
# Test warehouse robot agent
# Check robot actions in various warehouse scenarios
scenarios = [
    {"ObstacleAhead": True, "CarryingFragileItem": True, "HumanNearby": False, "PathClear": False, "DestinationReached": False},
    {"ObstacleAhead": False, "CarryingFragileItem": False, "HumanNearby": True, "PathClear": True, "DestinationReached": False},
    {"ObstacleAhead": False, "CarryingFragileItem": False, "HumanNearby": False, "PathClear": True, "DestinationReached": True},
]

for s in scenarios:
    action = warehouse_robot_agent(s)
    print(f"Percepts: {s} -> Action: {action}")
```
